```python
df = df.drop(columns=['img', 'last_update'])
```

```python
# 1. Define the columns that define a duplicate record
subset_cols = ['title', 'console', 'release_date']

# 2. Filter the dataframe to show only duplicate rows
duplicate_rows = df[df.duplicated(subset=subset_cols, keep=False)]

# 3. Sort by those columns so duplicate matches appear next to each other
duplicate_rows = duplicate_rows.sort_values(by=subset_cols)

# 4. Display the results
display(duplicate_rows)
```

```python
# 1. Sort by total_sales descending, pushing NaN values to the very end
df_sorted = df.sort_values(
    by='total_sales', 
    ascending=False, 
    na_position='last'
)

df = df_sorted.drop_duplicates(subset=['title', 'console', 'release_date'], keep='first')
```

```python
df
```

```python
# Standardize string columns
string_cols = ['title', 'publisher', 'developer', 'genre', 'console']
df[string_cols] = df[string_cols].apply(lambda col: col.str.strip())

# # Standardize in a single pipeline
# df['Clean_Company'] = (df['Company']
#                        .str.strip()               # Remove extra spaces on outer edges
#                        .str.lower()               # Enforce lower case for uniformity
#                        .str.replace(r'[^\w\s]', '', regex=True) # Strip punctuation
#                        .str.replace(r'\s+', ' ', regex=True))  # Collapse interior whitespace

df
```

```python
# datetime format
df['release_date'] = pd.to_datetime(df['release_date'], errors='coerce')

df
```

```python
# I'll just keep the NaN values empty in the numerical columns
```

```python
# Returns a True/False series
mask = df['developer'] == 'Unknown'

# Returns the actual rows matching the condition
filtered_df = df[mask]
# filtered_df = df[df['developer'].isnull()]

filtered_df
```

```python
df['developer'] = df['developer'].fillna('Unknown')

df
```

```python
calculated_total = df[['na_sales', 'jp_sales', 'pal_sales', 'other_sales']].sum(axis=1)
# Highlight rows where total sales deviates from regional sum
sales_discrepancies = df[abs(df['total_sales'] - calculated_total) > 0.05]

sales_discrepancies
```
